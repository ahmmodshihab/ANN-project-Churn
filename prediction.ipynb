```python
import tensorflow as tf
from tensorflow.keras.models import load_model
import pickle
import pandas as pd
import numpy as np
```

```python
model=load_model('model.h5')
```

```python
with open('gender_label_encoder.pkl','rb') as file:
    gender_label_encoder=pickle.load(file)

with open('geography_oh_encoder.pkl','rb') as file:
    geography_oh_encoder=pickle.load(file)
   
with open('scaler.pkl','rb') as file:
    scaler=pickle.load(file)
```

```python
# example input data
input_data = {
    'CreditScore': 600,
    'Geography': 'France',
    'Gender': 'Male',
    'Age': 40,
    'Tenure': 3,
    'Balance': 60000,
    'NumOfProducts': 2,
    'HasCrCard': 1,
    'IsActiveMember': 1,
    'EstimatedSalary': 50000
}
```

```python
input_data=pd.DataFrame([input_data])
```

```python
input_data
```

```python
input_data['Gender']=gender_label_encoder.transform(input_data['Gender'])
```

```python
input_data
```

```python
geo_transformed=geography_oh_encoder.transform(input_data[['Geography']]).toarray()
feature_name=geography_oh_encoder.get_feature_names_out(['Geography'])
geo_df=pd.DataFrame(geo_transformed,columns=feature_name)
```

```python
input_df=pd.concat([input_data.drop(['Geography'],axis=1),geo_df],axis=1)
```

```python
input_df
```

```python
input_scaled=scaler.transform(input_df)
```

```python
prediction=model.predict(input_scaled)
```

```python
prediction_proba = prediction[0][0]
```
